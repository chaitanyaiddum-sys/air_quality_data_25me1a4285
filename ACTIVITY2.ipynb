```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('air_quality.csv')

# --- DATA PREPARATION & AGGREGATION ---
# Filter data for just 'city' (matching the exact value in your CSV)
city_center_data = df[df['Location'] == 'city']

# Group data by location to find average pollution levels (using correct column name)
avg_pollution = df.groupby('Location')[['Air_Quality_Categerory', 'PM2.5_level']].agg(
    lambda x: x.mode()[0] if x.name == 'Air_Quality_Categerory' else x.mean()
)

# --- THE 5 REQUIRED CHARTS ---
# Chart 1: Line Chart (Tracking PM2.5_LEVEL over time at 'city')
plt.figure(figsize=(8, 4))
plt.plot(city_center_data['Date'],
         city_center_data['PM2.5_level'],
         marker='o', color='red')

plt.title('Chart 1: PM2.5 Level Trend at City')
plt.xlabel('Date')
plt.ylabel('PM2.5 Level')
plt.grid(True)
plt.show()
```

```python
# Chart 2: Bar Chart (Comparing average PM 2.5 by Location)

plt.figure(figsize=(8, 4))

avg_pollution['PM2.5_level'].plot(kind='bar', color='orange')

plt.title('Chart 2: Average PM2.5 Particles by Location')
plt.xlabel('Location')
plt.ylabel('Average PM2.5')
plt.xticks(rotation=0)

plt.show()
```

```python

# Chart 3: Scatter Plot (Distribution of PM2.5 levels over Date)

plt.figure(figsize=(8, 4))
plt.scatter(df['Date'], df['PM2.5_level'], color='purple', s=140)
plt.title('Chart 3: PM2.5 Level Distribution by Date')
plt.xlabel('Date')
plt.ylabel('PM2.5 level')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()

```

```python
#char 4
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/air_quality.csv')

plt.figure(figsize=(8, 4))
plt.hist(df['PM2.5_level'], bins=5, color='skyblue', edgecolor='black')
plt.title('char 4 :Distribution of PM2.5 levels')
plt.xlabel('PM2.5 level')
plt.ylabel('Frequency')
plt.show()
```

```python
# chart 5: Horizontal Bar chart (Average PM2.5 recorded per location)
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5_level'].mean().plot(kind='barh', color='blue')
plt.title('char 5 :Average PM2.5 levels by Location')
plt.xlabel('Average PM2.5 level')
plt.ylabel('Location')
plt.show()
```

```python
# Chart 5: Vertical Bar Chart (Max PM2.5 recorded per location)
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5_level'].max().plot(kind='bar', color='blue')
plt.title('Chart 5: Max PM2.5 Recorded per Location')
plt.xlabel('Location')
plt.ylabel('Max PM2.5')
plt.xticks(rotation=0)
plt.show()
```
